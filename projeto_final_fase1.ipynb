##Função Classificação NPS

```python
def nps_classificacao(df):
  if df['nps_score'] <= 6:
    return 'detratores'
  elif df['nps_score'] >= 9:
    return 'promotores'
  else:
    return 'neutros'
```

agora que classificamos o nps em classes. vamos trazer uma coluna :

```python
df_nps['classificacao'] = nps.apply(nps_classificacao, axis=1)
```

#Pairplot

```python
colunas = ['delivery_delay_days', 'complaints_count', 'customer_service_contacts',
           'resolution_time_days', 'csat_internal_score', 'repeat_purchase_30d']

df_nps.groupby('classificacao')[colunas].mean().round(2).reset_index()
```

Cada linha é o perfil médio de um grupo de clientes. Lendo da esquerda para a direita, detratores, neutros e promotores, quase todos os indicadores mudam de forma constante, em degraus:

Atraso na entrega: detratores esperaram em média 2,5 dias além do prazo, neutros 1,3 e promotores 0,7. Quanto maior o atraso, pior a nota. É o indicador que mais pesa na nota (correlação de -0,60).

Reclamações: detratores registraram 4,6 por cliente, contra 2,9 dos neutros e 2,3 dos promotores. Quem teve mais problemas reclama mais e avalia pior.

Contatos com o atendimento: detratores procuraram o atendimento 1,7 vez, e promotores 0,7. Precisar acionar o atendimento costuma sinalizar que algo deu errado.

Tempo de resolução: detratores esperaram 5,8 dias pela solução, e promotores 3,7. Resolver rápido ajuda, mas pesa menos que o atraso na entrega.

CSAT interno: sobe de 2,3 para 4,4 e 6,0. O CSAT e o NPS caminham juntos, então medem algo parecido.

Recompra em 30 dias: 0% dos detratores, 20% dos neutros e 100% dos promotores. Esse é o dado que dá argumento de negócio: cliente satisfeito volta a comprar. Mas 0% e 100% são valores "perfeitos" demais para dados reais, então vale comentar que a base provavelmente é simulada ou construída com essa regra.

Conclusão para a análise: o que separa um promotor de um detrator é a qualidade da experiência pós-compra (entrega no prazo, poucos problemas, atendimento rápido), e não o perfil do cliente nem o valor do pedido. Se a empresa quiser melhorar o NPS, o primeiro foco deveria ser reduzir os atrasos de entrega.

##1. Quais indicadores mais se relacionam com a nota? Mapa de calor da correlação

```python
colunas = ['nps_score', 'delivery_delay_days', 'complaints_count',
           'customer_service_contacts', 'resolution_time_days',
           'csat_internal_score', 'repeat_purchase_30d']

plt.figure(figsize=(8, 6))
sns.heatmap(df_nps[colunas].corr(), annot=True, fmt='.2f', cmap='RdYlGn', center=0)
plt.title('Correlação entre NPS e indicadores')
plt.show()
```

O mapa de calor mostra a correlação entre cada par de colunas, e o número vai de -1 a +1.

Como ler

Perto de +1 (verde): as duas sobem juntas.
Perto de -1 (vermelho): quando uma sobe, a outra desce.
Perto de 0 (amarelo): não há relação linear clara.
Como regra prática: acima de 0,5 em valor absoluto é forte, entre 0,3 e 0,5 é moderada e abaixo de 0,2 é fraca.
A diagonal é sempre 1,00, porque cada coluna é igual a ela mesma. Ignore.

O que olhar primeiro: a linha (ou coluna) nps_score

Indicador	Correlação com a nota	Leitura
Recompra em 30 dias	+0,57	Quem dá nota alta tende a recomprar
CSAT interno	+0,56	Os dois medem algo parecido
Atraso na entrega	-0,60	O mais forte: mais atraso, nota menor
Reclamações	-0,50	Mais reclamações, nota menor
Contatos com atendimento	-0,35	Efeito moderado
Tempo de resolução	-0,19	Efeito fraco

O que mais pesa na nota é a entrega atrasada. Depois vêm as reclamações.

Dois achados que aparecem fora da linha do NPS

Reclamações e contatos com atendimento têm correlação de 0,75. Quem reclama mais também liga mais para o atendimento, então os dois indicadores contam em parte a mesma história. Na análise, trate como um bloco só ("problemas e atendimento"), em vez de dois fatores independentes.
Tempo de resolução tem correlação de -0,74 com o CSAT, mas só -0,19 com o NPS. Resolver rápido melhora a satisfação interna, mas quase não muda a nota que o cliente dá. A nota depende mais de a entrega ter chegado no prazo.

Cuidados na interpretação

Correlação mostra associação, não causa provada. Dá para dizer "clientes com mais atraso tendem a dar notas menores".
A correlação mede só relação em linha reta. O gráfico de linha do atraso por nota (o terceiro que sugeri) completa essa leitura.
A recompra tem só valores 0 e 1 e é 100% nos promotores, então a correlação de +0,57 vem dessa regra "perfeita" da base e não deve ser tratada como achado de negócio forte.

Frase pronta para a apresentação: "O atraso na entrega é o principal fator associado à queda do NPS, seguido por reclamações. O tempo de resolução afeta o CSAT, mas tem pouco efeito sobre a nota."
